```python
import pandas as pd
df = pd.read_csv(r'C:\Users\uwmap\OneDrive\Documents\Usamah\Data-BI Analysis\Datasets\Marketing Campaign\marketing_campaign.csv', sep=';')
print(df.head())
```

```python
df_copy = df.copy()
df_copy.info()
```

```python
df_copy = df_copy.drop(['Z_CostContact', 'Z_Revenue'], axis=1)
df_copy.info()
```

```python
df_copy['TotalSpent'] = df_copy['MntWines'] + df_copy['MntFruits'] + df_copy['MntMeatProducts'] + df_copy['MntFishProducts'] + df_copy['MntSweetProducts'] + df_copy['MntGoldProds']
df_copy.head()
```

```python
df_copy = df_copy.drop(['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds'], axis=1)
```

```python

df_copy.drop(['Complain'], axis=1, inplace=True)
df_copy.info()
```

```python
df_copy 
```

```python
# Removing CMP columns, as most marketing campaigns were ignored by customers
df_copy.drop(['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5', 'Response'], axis=1, inplace=True)
df_copy 
```

```python
# checking for null values
df_copy.isnull().sum()
```

```python
# This will fill all NaN values in Income with the word 'Unknown'
df_copy['Income'] = df_copy['Income'].fillna('Unknown')
df_copy.isnull().sum()
```

```python
df_copy 
```

```python
# merge the kidhome and teenhome columns into a single column called Total_Children
df_copy['Total_Children'] = df_copy['Kidhome'] + df_copy['Teenhome']
df_copy = df_copy.drop(['Kidhome', 'Teenhome'], axis=1)
df_copy
```

```python
column_list = df_copy.columns.tolist()
print(column_list)
```

```python
# Reordering columns for better readability
new_column_order = ['ID', 'Year_Birth', 'Education', 'Marital_Status', 'Income', 'Total_Children', 'Dt_Customer', 'Recency', 'NumDealsPurchases', 'NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases', 'NumWebVisitsMonth', 'TotalSpent']
df_copy = df_copy[new_column_order] 
df_copy 
```

```python
#Change income DataType to int64
# Convert 'Unknown' values to 0 or the median income before converting to int64
df_copy['Income'] = df_copy['Income'].replace('Unknown', df_copy[df_copy['Income'] != 'Unknown']['Income'].astype('float64').median())
df_copy['Income'] = df_copy['Income'].astype('int64')
df_copy.info()
```

```python
df_copy.to_csv('cleaned_marketing_campaign1.csv', index=False)
```
